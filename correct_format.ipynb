### First, we correct the format of the voices

```python
def check_and_extract_metadata(directory):
    metadata_list = []
    incorrect_files = []
    for filename in os.listdir(directory):
        if filename.endswith(".mp3"):
            parts_underline = filename.split(".")[0].split("_")
            parts_dash = filename.split(".")[0].split("-")
            parts_dot = filename.split(".")[0].split(".")
            if(len(parts_underline) == 4):
                parts = parts_underline
            elif(len(parts_dash) == 4):
                parts = parts_dash
            elif(len(parts_dot) == 4):
                parts = parts_dot
            else:
                # print("Error: Incorrect file format for file: ", filename)
                incorrect_files.append(filename)
                continue
            # print(parts)   
            student_id = parts[2]
            question = parts[1]
            gender = parts[3]
            path = os.path.join(directory, filename)
            metadata_list.append({
                "filename": filename,
                "question_number": question,
                "student_id": student_id,
                "gender": gender,
                "path": path
            })
            # return pd.DataFrame(metadata_list)
    return pd.DataFrame(metadata_list) , incorrect_files

metadata_df, incorrect_files = check_and_extract_metadata(DATA_PATH)
metadata_df.head()
```
